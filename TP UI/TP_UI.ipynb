```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import requests
import pytest
import ipytest

def testBotonDeMensajeOK():

    driver = webdriver.Chrome()
    driver.get("https://centyc.cl/pruebas/elementos-web")
    driver.find_element(By.ID, 'btn-ok').click()

    assert driver.find_element(By.CLASS_NAME, "text-xl").text == "Operación exitosa"    

    driver.quit()

ipytest.run()
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pytest
import ipytest

ipytest.autoconfig()

def testBotonClickSotenido():
    driver = webdriver.Chrome()
    
    driver.get("https://centyc.cl/pruebas/elementos-web")
        
    boton = driver.find_element(By.ID, 'btn-sostenido')

    ActionChains(driver)\
        .click_and_hold(boton)\
        .pause(0.5)\
        .release()\
        .perform()

    contador = WebDriverWait(driver, 2).until(
        EC.visibility_of_element_located((By.CLASS_NAME, "ml-2"))
    )
        
    assert contador.text == "1"
        
    driver.quit()

ipytest.run()
```
