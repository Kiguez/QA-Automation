```python
def esFizz(unNumero) :
    return (unNumero % 3) == 0
```

```python
def esBuzz(unNumero) :
    return (unNumero % 5) == 0
```

```python
def esFizzBuzz(unNumero) :
    return esFizz(unNumero) and esBuzz(unNumero)
```

```python
def nrosDel1Al100ConFizzBuzz():
    for contador in range(1, 101) :
        salida = contador
        if(esFizzBuzz(contador)):
            salida = "FizzBuzz"
        elif(esFizz(contador)):
            salida = "Fizz"
        elif(esBuzz(contador)):
            salida = "Buzz"
        print(salida)
```
