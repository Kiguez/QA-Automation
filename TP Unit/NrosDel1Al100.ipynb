```python
def esFizz(unNumero) :
    return (unNumero % 3) == 0
```

```python
def esBuzz(unNumero) :
    return (unNumero % 5) == 0
```

```python
def esFizzBuzz(unNumero) :
    return esFizz(unNumero) and esBuzz(unNumero)
```

```python
def nrosConFizzBuzz(unNumero):
    salida = unNumero
    if(esFizzBuzz(unNumero)):
        salida = "FizzBuzz"
    elif(esFizz(unNumero)):
        salida = "Fizz"
    elif(esBuzz(unNumero)):
        salida = "Buzz"
    return salida
```

```python
def nrosDel1Al100ConFizzBuzz():
    salida = []
    for contador in range(1, 101) :
        salida.append(nrosConFizzBuzz(contador))
    return salida

print(nrosDel1Al100ConFizzBuzz())
```

```python
import unittest

class unoAlCienTest(unittest.TestCase):
    def testDadaUnaListaDel1Al100ConFizzBuzzCuandoLlamoAUnNumeroDivisiblePor3EntoncesMeDevelveFizz(self):
        actual = nrosDel1Al100ConFizzBuzz()[2]
        esperado = "Fizz"
        self.assertEqual(esperado , actual)

    def testDadaUnaListaDel1Al100ConFizzBuzzCuandoLlamoAUnNumeroDivisiblePor5EntoncesMeDevelveBuzz(self):
        actual = nrosDel1Al100ConFizzBuzz()[4]
        esperado = "Buzz"
        self.assertEqual(esperado , actual)

    def testDadaUnaListaDel1Al100ConFizzBuzzCuandoLlamoAUnNumeroDivisiblePor3Y5EntoncesMeDevelveFizzBuzz(self):
        actual = nrosDel1Al100ConFizzBuzz()[14]
        esperado = "FizzBuzz"
        self.assertEqual(esperado , actual)

    def testDadaUnaListaDel1Al100ConFizzBuzzCuandoLlamoAUnNumeroNoDivisiblePor3Y5EntoncesMeDevelveEseMismoNumero(self):
        actual = nrosDel1Al100ConFizzBuzz()[3]
        esperado = 4
        self.assertEqual(esperado , actual)

unittest.main(defaultTest="unoAlCienTest" , argv=[""] , verbosity=2 , exit=False)
```

```python

import pytest
import ipytest
ipytest.autoconfig()

def testDadaUnaFuncionEsFizzCuandoLePasoUnNumeroDivisiblePor3EntoncesEstaDevuelveVerdadero():
    assert esFizz(3) == True
    assert esFizz(9) == True
    assert esFizz(10) == False
    assert esFizz(5) == False


def testDadaUnaFuncionEsBuzzCuandoLePasoUnNumeroDivisiblePor5EntoncesEstaDevuelveVerdadero():
    assert esBuzz(5) == True
    assert esBuzz(10) == True
    assert esBuzz(3) == False
    assert esBuzz(9) == False

def testDadaUnaFuncionEsFizzBuzzCuandoLePasoUnNumeroDivisiblePor3Y5EntoncesEstaDevuelveVerdadero():
    assert esFizzBuzz(15) == True
    assert esFizzBuzz(30) == True
    assert esFizzBuzz(5) == False
    assert esFizzBuzz(3) == False

ipytest.run()

```
