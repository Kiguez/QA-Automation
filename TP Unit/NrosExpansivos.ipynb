```python
def siguienteDeNumeroExpansivo(expansivo):
    salida = []
    contador = 1
    for e in range (len(expansivo) - 1):
        actual = expansivo[e]
        siguiente = expansivo[e + 1]
        if(actual == siguiente):
            contador = contador + 1
        else:
            salida.append(contador)
            salida.append(actual)
            contador = 1
    salida.append(contador)
    salida.append(expansivo[-1])

    return salida

print(siguienteDeNumeroExpansivo([1 , 1]))
```
