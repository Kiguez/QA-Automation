```python
def expand(expansivo):
    salida = []
    contador = 1
    for e in range (len(expansivo) - 1):
        actual = expansivo[e]
        siguiente = expansivo[e + 1]
        contador = actualizarContadorYLista(contador , actual , siguiente , salida)
    salida.append(contador)
    salida.append(expansivo[- 1])

    return salida
```

```python
def actualizarContadorYLista(contador , actual , siguiente , lista):
    salida = contador
    if(actual == siguiente):
        salida = salida + 1
    else:
        lista.append(salida)
        lista.append(actual)
        salida = 1
    return salida
```

```python
import unittest

class expandTest(unittest.TestCase):
    def testDadoUnNumeroExpansivo1CuandoCalculoSuSiguienteEntoncesObtengo11(self):
        actual = expand([1])
        esperado = [1 , 1]
        self.assertEqual(esperado , actual)

    def testDadoUnNumeroExpansivo11CuandoCalculoSuSiguienteEntoncesObtengo21(self):
        actual = expand([1 , 1])
        esperado = [2 , 1]
        self.assertEqual(esperado , actual)

    def testDadoUnNumeroExpansivo21CuandoCalculoSuSiguienteEntoncesObtengo1211(self):
        actual = expand([2 , 1])
        esperado = [1 , 2 , 1 , 1]
        self.assertEqual(esperado , actual)

    def testDadoUnNumeroExpansivo1211CuandoCalculoSuSiguienteEntoncesObtengo111221(self):
        actual = expand([1 , 2 , 1 , 1])
        esperado = [1 , 1 , 1 , 2 , 2 , 1]
        self.assertEqual(esperado , actual)


unittest.main(defaultTest="expandTest" , argv=[""] , verbosity=2 , exit=False)
```

```python
def list2num(expansivo):
    salida = []
    for e in range(0 , (len(expansivo) - 1) , 2):
        actual = expansivo[e]
        siguiente = expansivo[e + 1]
        actualizarLista(actual , siguiente , salida)
    return salida
```

```python
def actualizarLista(actual, siguiente, lista):
    for n in range (actual):
        lista.append(siguiente)
```

```python
class list2numTest(unittest.TestCase):
    def testDadoUnNumeroExpansivo11CuandoCalculoSuVersionNormalEntoncesObtengo1(self):
        actual = list2num([1 , 1])
        esperado = [1]
        self.assertEqual(esperado , actual)

    def testDadoUnNumeroExpansivo21CuandoCalculoSuVersionNormalEntoncesObtengo11(self):
        actual = list2num([2 , 1])
        esperado = [1 , 1]
        self.assertEqual(esperado , actual)

    def testDadoUnNumeroExpansivo1211CuandoCalculoSuVersionNormalEntoncesObtengo21(self):
        actual = list2num([1 , 2 , 1 , 1])
        esperado = [2 , 1]
        self.assertEqual(esperado , actual)

    def testDadoUnNumeroExpansivo111221CuandoCalculoSuVersionNormalEntoncesObtengo1211(self):
        actual = list2num([1 , 1 , 1 , 2 , 2 , 1])
        esperado = [1 , 2 , 1 , 1]
        self.assertEqual(esperado , actual)


unittest.main(defaultTest="list2numTest" , argv=[""] , verbosity=2 , exit=False)
```
